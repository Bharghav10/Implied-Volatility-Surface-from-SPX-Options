```python
calls = selected_term[selected_term["option_type"] == "call"].copy()
puts = selected_term[selected_term["option_type"] == "put"].copy()
matched = calls.merge(
    puts,
    on=["quote_date","strike"],
    suffixes=("_call", "_put")
)

print("Matched calls and puts shape: ", len(matched))
matched.head()
```

```python
## For put-call parity analysis, we need to ensure that we have both call and put options for the same strike price and quote date. The matched DataFrame contains these pairs, allowing us to analyze the relationship between call and put prices effectively.
                        ## C - P = S - K*e^(-r*T)
##Where:
# C = Call option price
# P = Put option price
# S = Current stock price (underlying asset price)
# K = Strike price of the options
# r = Risk-free interest rate (annualized)
# For regression approch, we'll first calcul;ate:
            # C - P. and regresss against the strike price to see if the relationship holds. This will help us understand if the put-call parity condition is satisfied in the market data we have.
```

```python
matched["call_mid"] = (matched["bid_call"] + matched["ask_call"]) / 2
matched["put_mid"] = (matched["bid_put"] + matched["ask_put"]) / 2
matched["call_minus_put"] = matched["call_mid"] - matched["put_mid"]
parity_data = matched[["quote_date", "strike", "underlying_last_call", "dte_call", "call_mid", "put_mid", "call_minus_put"]].copy()
parity_data.head()
```

```python
import numpy as np
import statsmodels.api as sm
regression_results = []

regression_results = []
for quote_date, group in parity_data.groupby("quote_date"):
    X = sm.add_constant(group["strike"])  # Add a constant term for the intercept
    y = group["call_minus_put"]
    model = sm.OLS(y, X).fit()

    intercept = model.params["const"]
    slope = model.params["strike"]

    S = group["underlying_last_call"].iloc[0]  # Current stock price
    T = group["dte_call"].iloc[0] / 365  # Time to expiration in years

    if slope <0:
        r = -np.log(-slope) / T
    else:
        r = np.nan  # If slope is not negative, we cannot compute a valid risk-free rate

    regression_results.append({
        "quote_date": quote_date,
        "n_pairs": len(group),
        "underlying": S,
        "dte": group["dte_call"].iloc[0],
        "intercept": intercept,
        "slope": slope,
        "r_decimal": r,
        "r_percent": r * 100,
        "r_squared": model.rsquared,
        "slope_tstat": model.tvalues["strike"],
        "slope_pvalue": model.pvalues["strike"],
        "slope_std_error": model.bse["strike"]

    })

regression_df = pd.DataFrame(regression_results)
regression_df
```

```python
# There is a negative regression slope on 14 of November 2023, which indicates that with only 3 days to expiration pricing/bid-ask effects can have effect on calculated rate.
```

```python
comparison = regression_df[
    ["quote_date", "n_pairs", "underlying", "dte", "r_percent", "r_squared"]
].copy()
comparison = comparison.merge(
    cme_rates[["quote_date", "cme_rates_pct"]],
    on="quote_date",
    how="left")
comparison["difference_pct_points"] = (
comparison["r_percent"] - comparison["cme_rates_pct"]
)

comparison 
```

```python
# Save the regression results to a CSV file
regression_df.to_csv("../results/parity_regression.csv", index=False)

# Save Cme vs Parity comparison
comparison.to_csv("../results/cme_vs_parity.csv", index=False)

print("Saved: ")
print("../results/parity_regression.csv")
print("../results/cme_vs_parity.csv")
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.plot(comparison["quote_date"], comparison["r_percent"], marker='o', label="Put-Call Parity")
plt.plot(comparison["quote_date"], comparison["cme_rates_pct"], marker='o', label="CME SR1X3")

plt.axhline(0, linewidth=0.8)
plt.xlabel("Quote Date")
plt.ylabel("Interest Rate (%)")
plt.title("SPX Implied Risk-free rate vs Cme SR1X3 Rates")
plt.legend()
plt.grid(True, alpha = 0.3)

plt.show()
```
