```python
import pandas as pd 
```

```python
q1 = pd.read_csv('../data/Q1/max_spx_options_filtered_2023-01-04.csv')
q2 = pd.read_csv('../data/Q2/max_spx_options_filtered_2023-06-30.csv')
q3 = pd.read_csv('../data/Q3/max_spx_options_filtered_2023-08-15.csv')
q4 = pd.read_csv('../data/Q4/max_spx_options_filtered_2023-11-14.csv')
print(q1.shape)
print(q2.shape)
print(q3.shape)
print(q4.shape)
```

```python
all_data = pd.concat([q1, q2, q3, q4], ignore_index=True)
print("Combined shape: ", all_data.shape)
```

```python
q1["quarter"] = "Q1"
q2["quarter"] = "Q2"
q3["quarter"] = "Q3"
q4["quarter"] = "Q4"

all_data = pd.concat([q1, q2, q3, q4], ignore_index=True)

print(all_data.shape)
```

```python
expiry_counts = all_data["expire_date"].value_counts()

expiry_counts_df = (
    expiry_counts
    .rename_axis("expire_date")
    .reset_index(name="option_count")
)

expiry_counts_df.head(20)

expiry_counts_df.head(20)
```

```python
top_term = expiry_counts_df.iloc[0]

print("Term with most options: ")
print("Expiry:", top_term["expire_date"])
print("Number of options:", top_term["option_count"])
```

```python
all_data[all_data["expire_date"] == top_term["expire_date"]].head(20)
all_data[all_data["expire_date"] == top_term["expire_date"]].shape
all_data[all_data["expire_date"] == top_term["expire_date"]]["quarter"].value_counts()
all_data[all_data["expire_date"] == top_term["expire_date"]]["option_type"].value_counts()
```

```python
selected_term = all_data[all_data["expire_date"] == top_term["expire_date"]]
selected_term[["dte"]].describe()

selected_term[["expire_date", "dte"]].head(20)
```

```python
import os

os.makedirs("../results", exist_ok=True)
expiry_counts_df.to_csv(
    "../results/expiry_counts.csv",
    index=False
)
print("Saved: results/expiry_counts.csv")
```

```python
selected_term.to_csv(
    "../results/selected_term.csv",
    index=False
)
print("Saved: results/selected_term_options.csv")
```

```python

selected_term["quote_date"].value_counts().sort_index()
selected_term.groupby("quote_date")["dte"].first()
```

```python
selected_term[
    ["quote_date", "expire_date", "dte"]
].drop_duplicates().sort_values("quote_date")
```

```python
cme_dates = (
    selected_term[
    ["quote_date", "expire_date", "underlying_last", "dte"]
].drop_duplicates().sort_values("quote_date").reset_index(drop=True)
)
cme_dates["cme_contract"] = "SR1X3" 
cme_dates 
```

```python
cme_rates = pd.DataFrame({
    "quote_date": [
        "2023-01-04",
        "2023-06-30",
        "2023-08-15",
        "2023-11-14"
    ],
    "expire_date": [
        "2023-11-17",
        "2023-11-17",
        "2023-11-17",
        "2023-11-17"
    ],
    "cme_contract": [
        "SR1X3",
        "SR1X3",
        "SR1X3",
        "SR1X3"
    ],
    "settlement":[
        95.255,
        94.595,
        94.590,
        94.685
    ]
})
cme_rates["cme_rates_pct"] = 100 - cme_rates["settlement"]
cme_rates 
```

```python
os.makedirs("../results", exist_ok=True)
cme_rates.to_csv(
    "../results/cme_rates.csv",
    index=False
)
print("CME rates saved successfully.")
```

```python
selected_term_with_cme = selected_term.merge(
    cme_rates[["quote_date", "cme_contract","settlement", "cme_rates_pct"]],
    on="quote_date",
    how="left"
)   
selected_term_with_cme.head()
```

```python
selected_term_with_cme[
    ["quote_date", "expire_date", "cme_contract", "settlement", "cme_rates_pct"]
].drop_duplicates().sort_values("quote_date")
```

```python
selected_term_with_cme.to_csv(
    "../results/selected_term_with_cme.csv",
    index=False
)
print("Combined SPX and Cme date saved successfully.")
```
